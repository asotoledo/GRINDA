## Read Data

```python
# Load detailed disruption simulation results (steps) for GAT using pickle
with open('data/gat_steps_disrupt.pkl', 'rb') as f:
    gat_steps_disrupt = pickle.load(f)

# Load detailed disruption simulation results (steps) for GT using pickle
with open('data/gt_steps_disrupt.pkl', 'rb') as f:
    gt_steps_disrupt = pickle.load(f)

# Load summary results for GAT from CSV
gat_summary_disrupt = pd.read_csv('data/gat_summary_disrupt.csv')

# Load summary results for GT from CSV
gt_summary_disrupt = pd.read_csv('data/gt_summary_disrupt.csv')

# Load the dictionaries of DataFrames directly from pickle files
with open('data/gat_outlier.pkl', 'rb') as f:
    gat_scores_dfs = pickle.load(f)  # Dictionary: {network_id: {channels: DataFrame}}

with open('data/gt_outlier.pkl', 'rb') as f:
    gt_scores_dfs = pickle.load(f)  # Dictionary: {network_id: {channels: DataFrame}}
```

## Results 

```python
# Selects the best model per network and channels according to the defined criteria:
# 1. Highest 'number_of_components'
# 2. Lowest 'global_efficiency'
# 3. Lowest 'largest_component_nodes'
gat_best_models = (
    gat_summary_disrupt
    .sort_values(['network', 'channels', 'number_of_components', 'global_efficiency', 'largest_component_nodes'],
                 ascending=[True, True, False, True, True])
    .groupby(['network', 'channels'], as_index=False)
    .first()
)

gat_best_models_by_network = (
    gat_best_models
    .sort_values(['network', 'steps_to_disruption', 'number_of_components', 'global_efficiency', 'largest_component_nodes'],
                 ascending=[True, True, False, True, True])
    .groupby('network', as_index=False)
    .first()
)

gt_best_models = (
    gt_summary_disrupt
    .sort_values(['network', 'channels', 'number_of_components', 'global_efficiency', 'largest_component_nodes'],
                 ascending=[True, True, False, True, True])
    .groupby(['network', 'channels'], as_index=False)
    .first()
)

gt_best_models_by_network = (
    gt_best_models
    .sort_values(['network', 'steps_to_disruption', 'number_of_components', 'global_efficiency', 'largest_component_nodes'],
                 ascending=[True, True, False, True, True])
    .groupby('network', as_index=False)
    .first()
)
```

```python
# Copy the best models by network for GAT and add a column indicating the graph type
gat_best = gat_best_models_by_network.copy()
gat_best['graph'] = 'GAT'

# Copy the best models by network for GT and add a column indicating the graph type
gt_best = gt_best_models_by_network.copy()
gt_best['graph'] = 'GT'

# Concatenate GAT and GT best models into a single DataFrame
combined = pd.concat([gat_best, gt_best], ignore_index=True)

# Add the 'num_nodes' column with the number of nodes for each network
nodes_count = []
for _, row in combined.iterrows():
    graph = row['graph']
    network = row['network']
    channels = row['channels']
    # Select the correct scores dictionary based on graph type
    scores_dict = gat_scores_dfs if graph == 'GAT' else gt_scores_dfs
    scores_df = scores_dict.get(network, {}).get(channels, None)
    if scores_df is not None:
        nodes_count.append(len(scores_df))
    else:
        nodes_count.append(np.nan)
combined['num_nodes'] = nodes_count

# Sort by number of nodes (descending)
combined_sorted = combined.sort_values(
    ['num_nodes', 'network', 'steps_to_disruption', 'number_of_components', 'global_efficiency', 'largest_component_nodes'],
    ascending=[False, True, True, False, True, True]
)

# Select the best model per network based on the criteria, but keep the order by num_nodes
best_models_by_network = (
    combined_sorted
    .groupby('network', as_index=False)
    .first()
    .sort_values('num_nodes', ascending=False)  # Final sorting by num_nodes
    .loc[:, ['graph', 'num_nodes'] + [col for col in gat_best_models_by_network.columns]]
)

# Save the resulting DataFrame to CSV
best_models_by_network.to_csv('data/best_models_by_network.csv', sep=';', index=False)

print("Plotting Best Models for network (ordered by number of nodes)...")
display(best_models_by_network)
```

## Plot Disruption

```python
# Converts the detailed disruption results dictionary into a dictionary of DataFrames
def steps_disrupt_to_dfs(steps_disrupt):
    """
    Converts a dictionary of detailed disruption results into a dictionary of DataFrames.
    Each key (network_id, channels, model) will have a DataFrame with the steps and metrics.
    """
    disruption_dfs = {}
    for key, values in steps_disrupt.items():
        # Ensure all values are lists or arrays of the same length for DataFrame construction
        df = pd.DataFrame({
            k: list(v) if isinstance(v, (np.ndarray, list)) else [v]*len(next(iter(values.values())))
            for k, v in values.items()
        })
        disruption_dfs[key] = df
    return disruption_dfs

# Convert GAT and GT disruption results to DataFrame dictionaries
gat_disrupt_dfs = steps_disrupt_to_dfs(gat_steps_disrupt)
gt_disrupt_dfs = steps_disrupt_to_dfs(gt_steps_disrupt)
```

```python
def get_axis_domain(axis, idx):
    # Returns the axis domain string for plotly subplots.
    # axis: 'x' or 'y', idx: subplot index (1-based)
    return f"{axis} domain" if idx == 1 else f"{axis}{idx} domain"

class DisruptionPlotterPlotly:
    def __init__(self, network_id, colors=None, linestyles=None):
        # Initialize the plotter with network_id, colors, and linestyles.
        self.network_id = network_id
        self.colors = colors or [
            '#66c2a5', '#fc8d62', '#8da0cb', '#e78ac3', 
            '#a6d854', '#ffd92f', '#e5c494', '#b3b3b3'
        ]
        self.linestyles = linestyles or [
            'solid', 'dash', 'dot', 'dashdot', 'longdash', 'longdashdot'
        ]

    def plot_comparative(self, disrupt_dfs, best_models, best_models_by_network, axes_col, fig, col_title):
        # Plots comparative disruption curves for a given network and column (GAT/GT).
        rows = best_models[best_models['network'] == self.network_id]
        best_row = best_models_by_network[best_models_by_network['network'] == self.network_id]
        best_key = None
        best_color = None
        best_steps = None
        best_global_eff = None
        best_largest_comp = None
        if not best_row.empty:
            # Extract best model info for annotation and vertical line.
            best_row = best_row.iloc[0]
            best_key = (best_row['network'], best_row['channels'], best_row['model'])
            best_steps = best_row['steps_to_disruption']
            best_global_eff = best_row['global_efficiency']
            best_largest_comp = best_row['largest_component_nodes']

        labels = []
        for i, row in enumerate(rows.itertuples()):
            key = (row.network, row.channels, row.model)
            if key not in disrupt_dfs:
                continue
            df = disrupt_dfs[key]
            color = self.colors[i % len(self.colors)]
            dash = self.linestyles[i % len(self.linestyles)]
            label = f"{row.model}/{row.channels}"

            # Set hover template and legend visibility for each column.
            if axes_col == 0:
                hovertemplate = (
                    f"<b>{label}</b><br>"
                    "Step: %{x}<br>"
                    "Value: %{y:.3f}<extra></extra>"
                )
                showlegend = False
            else:
                hovertemplate = (
                    f"<b>{label}</b><br>"
                    "Step: %{x}<br>"
                    "Value: %{y:.3f}<extra></extra>"
                )
                showlegend = False

            # Add normalized components curve to subplot row 1.
            fig.add_trace(
                go.Scatter(
                    x=df['steps'], y=df['normalized_components'],
                    mode='lines',
                    name=None,
                    line=dict(color=color, width=3, dash=dash),
                    legendgroup=label,
                    showlegend=showlegend,
                    hovertemplate=hovertemplate
                ),
                row=1, col=axes_col+1
            )
            # Add normalized global efficiency curve to subplot row 2.
            fig.add_trace(
                go.Scatter(
                    x=df['steps'], y=df['normalized_global_efficiency'],
                    mode='lines',
                    name=None,
                    line=dict(color=color, width=3, dash=dash),
                    legendgroup=label,
                    showlegend=showlegend,
                    hovertemplate=hovertemplate
                ),
                row=2, col=axes_col+1
            )
            # Add normalized largest component size curve to subplot row 3.
            fig.add_trace(
                go.Scatter(
                    x=df['steps'], y=df['normalized_size_largest_component'],
                    mode='lines',
                    name=None,
                    line=dict(color=color, width=3, dash=dash),
                    legendgroup=label,
                    showlegend=showlegend,
                    hovertemplate=hovertemplate
                ),
                row=3, col=axes_col+1
            )
            if key == best_key:
                # Store color for best model annotation and vertical line.
                best_color = color

        # Add vertical lines at disruption step for best model.
        if best_steps is not None and best_color is not None:
            for idx in range(1, 4):
                fig.add_vline(
                    x=best_steps, line_dash="dash", line_color=best_color, line_width=3,
                    row=idx, col=axes_col+1
                )
            # Add annotation marker for disruption step in first row.
            fig.add_trace(
                go.Scatter(
                    x=[best_steps],
                    y=[0.2],
                    mode="markers+text",
                    marker=dict(color=best_color, size=12, symbol="line-ns-open"),
                    text=[f"Disruption: Step {best_steps}"],
                    textposition="top right",
                    textfont=dict(size=20, color=best_color, family="serif"),
                    hovertemplate=f"Disruption: Step {best_steps}<extra></extra>",
                    showlegend=False
                ),
                row=1, col=axes_col+1
            )

        # Set axis labels and formatting for each subplot row.
        ylabels = [
            'Components (normalized)', 
            'Global Efficiency (normalized)', 
            'Largest Component Size (normalized)'
        ]
        for idx in range(1, 4):
            fig.update_yaxes(title_text=ylabels[idx-1] if axes_col == 0 else "", row=idx, col=axes_col+1)
            fig.update_xaxes(title_text="Steps" if idx == 3 else "", row=idx, col=axes_col+1)
            if idx == 1:
                fig.update_xaxes(showticklabels=False, row=idx, col=axes_col+1)
                fig.update_yaxes(title_font=dict(size=20), row=idx, col=axes_col+1)
                fig.update_yaxes(tickfont=dict(size=20), row=idx, col=axes_col+1)
                fig.update_xaxes(tickfont=dict(size=20), row=idx, col=axes_col+1)
            elif idx < 3:
                fig.update_xaxes(showticklabels=False, row=idx, col=axes_col+1)
        # Add column title annotation above the subplots.
        xref = get_axis_domain('x', axes_col+1)
        fig.add_annotation(
            text=f"<b>{col_title}</b>",
            xref=xref,
            yref="paper",
            x=0.5, y=1.08,
            showarrow=False,
            font=dict(size=22),
            row=1, col=axes_col+1
        )

    def plot_all(self, gat_steps_disrupt, gat_best_models, gat_best_models_by_network,
                 gt_steps_disrupt, gt_best_models, gt_best_models_by_network):
        # Plots all comparative curves for both GAT and GT networks.
        fig = make_subplots(
            rows=3, cols=2,
            shared_xaxes=False,
            vertical_spacing=0.07,
            horizontal_spacing=0.08,
            subplot_titles=("", "", "", "", "", "")
        )
        # Plot GAT curves in first column.
        self.plot_comparative(gat_steps_disrupt, gat_best_models, gat_best_models_by_network, 0, fig, "GAT")
        # Plot GT curves in second column.
        self.plot_comparative(gt_steps_disrupt, gt_best_models, gt_best_models_by_network, 1, fig, "GT")

        # Update layout and show the figure.
        fig.update_layout(
            height=1200, width=1400,
            showlegend=False,
            font=dict(family="serif", size=20),
            plot_bgcolor='white',
            paper_bgcolor='white',
            margin=dict(t=80, b=120, l=40, r=40)
        )
        fig.show()
```

## Key Roles Identifing

```python
class KeyRolesPlotterPlotly:
    def __init__(self, best_models_by_network, gat_scores_dfs, gt_scores_dfs):
        # Store input dataframes and dictionaries
        self.best_models_by_network = best_models_by_network
        self.gat_scores_dfs = gat_scores_dfs
        self.gt_scores_dfs = gt_scores_dfs
        self.network_scores_data = {}

    def collect_scores(self, networks_to_plot=None):
        # Collect scores for each network to plot
        for _, row in self.best_models_by_network.iterrows():
            network = row['network']
            # Skip networks not in the selected list
            if networks_to_plot is not None and network not in networks_to_plot:
                continue
            graph = row['graph']
            channels = row['channels']
            model = row['model']
            steps_to_disruption = row['steps_to_disruption']
            # Select the correct scores dictionary based on graph type
            scores_dict = self.gat_scores_dfs if graph == 'GAT' else self.gt_scores_dfs
            scores_df = scores_dict.get(network, {}).get(channels, None)
            if scores_df is None:
                continue
            score_key = f'score_{model}'
            # Skip if the score column is missing
            if score_key not in scores_df.columns:
                continue
            # Prepare the data for plotting
            data = scores_df[['node_id', score_key]].rename(columns={'node_id': 'id', score_key: 'score'})
            self.network_scores_data[network] = {
                'data': data,
                'steps_to_disruption': steps_to_disruption,
                'model': model
            }

    def plot(self, networks_to_plot):
        from plotly.subplots import make_subplots
        import plotly.graph_objects as go

        # Define color palette for plotting
        palette = ["#636EFA", "#AB63FA"]
        n_networks = len(networks_to_plot)
        max_spacing = 1 / max(n_networks - 1, 1)
        # Create subplots for each network
        fig = make_subplots(
            rows=n_networks, cols=1,
            subplot_titles=["" for _ in networks_to_plot],
            vertical_spacing=min(0.02, max_spacing)
        )
        for idx, network in enumerate(networks_to_plot):
            info = self.network_scores_data.get(network)
            if info is None:
                continue
            data = info['data']
            steps_to_disruption = info['steps_to_disruption']
            model = info['model']
            # Select top nodes as key roles
            top_nodes = data.nlargest(steps_to_disruption, 'score')
            top_ids = set(top_nodes['id'])

            node_indices = data.index + 1

            # Plot non-key roles (other nodes)
            mask_non_key = ~data['id'].isin(top_ids)
            fig.add_trace(
                go.Scatter(
                    x=node_indices[mask_non_key],
                    y=data[mask_non_key]['score'],
                    mode='markers',
                    marker=dict(color=palette[0], size=8, line=dict(width=1, color='black')),
                    name="Other nodes",
                    legendgroup="nodes",
                    showlegend=(idx == 0),
                    hovertemplate="Node: %{customdata}<br>Score: %{y:.4f}<extra></extra>",
                    customdata=data[mask_non_key]['id']
                ),
                row=idx+1, col=1
            )
            # Plot key roles
            mask_key = data['id'].isin(top_ids)
            fig.add_trace(
                go.Scatter(
                    x=node_indices[mask_key],
                    y=data[mask_key]['score'],
                    mode='markers',
                    marker=dict(color=palette[1], size=14, line=dict(width=1, color='black')),
                    name="Key roles",
                    legendgroup="keyroles",
                    showlegend=(idx == 0),
                    hovertemplate="Key Role: %{customdata}<br>Score: %{y:.4f}<extra></extra>",
                    customdata=data[mask_key]['id']
                ),
                row=idx+1, col=1
            )
            # Plot cutoff line for key roles
            min_top_nodes_value = top_nodes['score'].min()
            cutoff_y = min_top_nodes_value - 0.01 * min_top_nodes_value
            fig.add_trace(
                go.Scatter(
                    x=[1, len(data)],
                    y=[cutoff_y, cutoff_y],
                    mode='lines',
                    line=dict(color="#00CC96", dash='dash', width=3),
                    name="Cutoff",
                    legendgroup="cutoff",
                    showlegend=(idx == 0),
                    hoverinfo="skip"
                ),
                row=idx+1, col=1
            )
            # Add vertical y-axis line at x=1
            fig.add_shape(
                type="line",
                x0=1, x1=1,
                y0=data['score'].min() - abs(data['score'].min()) * 0.01,
                y1=data['score'].max() + abs(data['score'].max()) * 0.01,
                line=dict(color="black", width=2, dash="dot"),
                row=idx+1, col=1
            )
            # Add horizontal x-axis line at y=0
            fig.add_shape(
                type="line",
                x0=1, x1=len(data),
                y0=0, y1=0,
                line=dict(color="black", width=2, dash="dot"),
                row=idx+1, col=1
            )

            # Set y-axis range with margin
            y_margin = (data['score'].max() - data['score'].min()) * 0.03
            fig.update_yaxes(
                title_text=f"{model} score",
                range=[data['score'].min() - y_margin, data['score'].max() + y_margin],
                row=idx+1, col=1
            )

            # Set x-axis range with margin and label
            x_margin = int(len(data) * 0.03)
            fig.update_xaxes(
                title_text=f"Node index – Network {network}",
                range=[1-x_margin, len(data)+x_margin],
                row=idx+1, col=1
            )

        # Remove subplot titles (extra guarantee)
        if "annotations" in fig['layout']:
            fig['layout']['annotations'] = []

        # Update overall figure layout
        fig.update_layout(
            height=700*n_networks,
            width=900,
            title_text="",
            margin=dict(t=60),
            legend=dict(
                orientation="h",
                yanchor="bottom",
                y=-0.18,
                xanchor="center",
                x=0.5,
                font=dict(size=15)
            ),
            font=dict(family="serif", size=16),
            plot_bgcolor='white',
            paper_bgcolor='white'
        )
        fig.show()
```

```python
# ======== Classes ========

class NetworkDisruptor:
    def __init__(self, network_number):
        self.network_number = network_number
        self.graph_type = None
        self.channels = None
        self.model = None
        self.steps_to_disruption = None
        self.ranking_df = None
        self.edgelist_df = None
        self.G_full = None
        self.plotter = None

    def load_best_model_params(self, csv_path='data/best_models_by_network.csv'):
        df = pd.read_csv(csv_path, sep=';')
        row = df[df['network'] == self.network_number].iloc[0]
        self.graph_type = row['graph']
        self.channels = int(row['channels'])
        self.model = row['model']
        self.steps_to_disruption = int(row['steps_to_disruption'])

    def load_ranking(self, pkl_path='data/gt_ranking.pkl'):
        with open(pkl_path, 'rb') as f:
            ranking_tables = pickle.load(f)
        self.ranking_df = ranking_tables[self.network_number][self.channels][self.model]

    def load_edgelist(self):
        edgelist_path = f'data/network_{self.network_number}.edgelist'
        self.edgelist_df = pd.read_csv(edgelist_path, delim_whitespace=True, header=None, names=['Source', 'Target', 'Attr'])
        self.edgelist_df = self.edgelist_df[['Source', 'Target']]

    def build_graph(self):
        nodes_in_source = set(self.edgelist_df['Source'].dropna())
        nodes_in_target = set(self.edgelist_df['Target'].dropna())
        all_nodes = nodes_in_source.union(nodes_in_target)
        isolated_nodes = set()
        if self.edgelist_df['Source'].isnull().any():
            isolated_nodes.update(self.edgelist_df.loc[self.edgelist_df['Source'].isnull(), 'Target'].dropna())
        if self.edgelist_df['Target'].isnull().any():
            isolated_nodes.update(self.edgelist_df.loc[self.edgelist_df['Target'].isnull(), 'Source'].dropna())
        all_nodes = all_nodes.union(isolated_nodes)
        G = nx.DiGraph()
        G.add_nodes_from(all_nodes)
        edges = self.edgelist_df.dropna(subset=['Source', 'Target']).values.tolist()
        G.add_edges_from(edges)
        self.G_full = G

    def disrupt_network(self):
        for step in range(self.steps_to_disruption):
            if step >= len(self.ranking_df):
                break
            node_to_remove = self.ranking_df.iloc[step]
            if node_to_remove in self.G_full:
                self.G_full.remove_node(node_to_remove)

class Network3DPlotter:
    def __init__(self, G):
        self.G = G
        self.pos = None

    def compute_layout(self, k=30, iterations=1000, seed=42):
        self.pos = nx.spring_layout(self.G, dim=3, seed=seed, k=k, iterations=iterations)

    def plot(self, title="Network Graph"):
        Xn = [self.pos[k][0] for k in self.G.nodes()]
        Yn = [self.pos[k][1] for k in self.G.nodes()]
        Zn = [self.pos[k][2] for k in self.G.nodes()]
        labels = [str(n) for n in self.G.nodes()]
        Xe, Ye, Ze = [], [], []
        for e in self.G.edges():
            Xe += [self.pos[e[0]][0], self.pos[e[1]][0], None]
            Ye += [self.pos[e[0]][1], self.pos[e[1]][1], None]
            Ze += [self.pos[e[0]][2], self.pos[e[1]][2], None]
        fig = go.Figure()
        fig.add_trace(go.Scatter3d(
            x=Xe, y=Ye, z=Ze,
            mode='lines',
            line=dict(color='gray', width=2),
            hoverinfo='none'
        ))
        fig.add_trace(go.Scatter3d(
            x=Xn, y=Yn, z=Zn,
            mode='markers+text',
            marker=dict(
                symbol='circle',
                size=20,
                color='rgba(142,202,230,0.6)',
                line=dict(color='rgba(2,48,71,1)', width=3),
                opacity=0.7
            ),
            text=labels,
            textposition="middle center",
            textfont=dict(size=13, color='black'),
            hoverinfo='text'
        ))
        fig.update_layout(
            title=title,
            showlegend=False,
            scene=dict(
                xaxis=dict(showbackground=False, showticklabels=False, visible=False),
                yaxis=dict(showbackground=False, showticklabels=False, visible=False),
                zaxis=dict(showbackground=False, showticklabels=False, visible=False),
            ),
            margin=dict(l=0, r=0, b=0, t=40),
            width=900,
            height=700,
            paper_bgcolor='rgba(0,0,0,0)',
            plot_bgcolor='rgba(0,0,0,0)'
        )
        fig.show()

# ======== Utility to list networks ========

def list_networks():
    arquivos = os.listdir('data')
    redes = []
    for nome in arquivos:
        if nome.startswith('network_') and nome.endswith('.edgelist'):
            try:
                num = int(nome.replace('network_', '').replace('.edgelist', ''))
                redes.append(num)
            except:
                pass
    redes.sort()
    return redes

# ======== DASHBOARD WIDGETS ========

available_networks = list_networks()
if not available_networks:
    available_networks = [1]

network_number_widget = widgets.Dropdown(
    options=[('', None)] + [(str(n), n) for n in available_networks],
    value=None,
    description='Rede:'
)
state_radio_widget = widgets.ToggleButtons(
    options=['Before Disruption', 'After Disruption'],
    value='Before Disruption',
    description='State:'
)
output = widgets.Output()

def update_dashboard(change=None):
    with output:
        clear_output(wait=True)
        if network_number_widget.value is None:
            print("Selecione uma rede para visualizar.")
            return
        try:
            network_number = network_number_widget.value
            estado = state_radio_widget.value
            disruptor = NetworkDisruptor(network_number)
            disruptor.load_best_model_params()
            disruptor.load_ranking()
            disruptor.load_edgelist()
            disruptor.build_graph()
            if estado == 'After Disruption':
                disruptor.disrupt_network()
            G = disruptor.G_full
            total_nodes = G.number_of_nodes()
            largest_component_nodes = max(nx.weakly_connected_components(G), key=len)
            num_largest_component_nodes = len(largest_component_nodes)
            #componentes = total_nodes - num_largest_component_nodes
            component = nx.number_weakly_connected_components(G)
            isolates = list(nx.isolates(G))
            print(f"Total Nodes: {total_nodes}")
            print(f"Largest Component Nodes: {num_largest_component_nodes}")
            print(f"Components: {component}")
            print(f"Isolated Nodes: {isolates[:10]}{' ...' if len(isolates) > 10 else ''}")
            plotter = Network3DPlotter(G)
            plotter.compute_layout(k=30, iterations=1000, seed=42)
            plotter.plot(title=f"Network {network_number} ({estado})")
        except Exception as e:
            print(f"Error loading network {network_number}: {e}")

# Avoid multiple observers in Voila
if not hasattr(update_dashboard, "_observer_set"):
    network_number_widget.observe(update_dashboard, names='value')
    state_radio_widget.observe(update_dashboard, names='value')
    update_dashboard._observer_set = True

display(widgets.HTML("<h2>GRINDA - Network Disruption Dashboard</h2>"))
display(widgets.HBox([network_number_widget, state_radio_widget]))
display(output)
```

```python
class KeyRoleLabeler:
    def __init__(self, individuals_df, key_roles_dict, best_models_by_network, gat_scores_dfs, gt_scores_dfs):
        """
        individuals_df: DataFrame with individuals in large components
        key_roles_dict: {network_id: [key_role_ids]}
        best_models_by_network: DataFrame with best model info per network
        gat_scores_dfs, gt_scores_dfs: dicts with scores per network/channel
        """
        self.individuals_df = individuals_df.copy()
        self.key_roles_dict = key_roles_dict
        self.best_models_by_network = best_models_by_network
        self.gat_scores_dfs = gat_scores_dfs
        self.gt_scores_dfs = gt_scores_dfs

    def label_key_roles(self):
        # Initialize 'key' column with 0
        self.individuals_df['key'] = 0
        # Mark key roles for each network
        for network_id, key_roles in self.key_roles_dict.items():
            key_roles_set = set(key_roles)
            mask = self.individuals_df['id'].isin(key_roles_set)
            self.individuals_df.loc[mask, 'key'] = 1

    def add_scores(self):
        # Initialize 'scores' column with NaN
        self.individuals_df['scores'] = np.nan
        # Add scores from the best model for each network
        for _, row in self.best_models_by_network.iterrows():
            network_id = row['network']
            graph = row['graph']
            channels = row['channels']
            model = row['model']
            scores_dict = self.gat_scores_dfs if graph == 'GAT' else self.gt_scores_dfs
            scores_df = scores_dict.get(network_id, {}).get(channels, None)
            if scores_df is None:
                continue
            score_key = f'score_{model}'
            if score_key not in scores_df.columns:
                continue
            aux = scores_df[['node_id', score_key]].rename(columns={'node_id': 'id', score_key: 'scores'})
            mask = self.individuals_df['id'].isin(aux['id'])
            self.individuals_df.loc[mask, 'scores'] = self.individuals_df.loc[mask].merge(
                aux, on='id', how='left')['scores_y'].values

    def process(self):
        self.label_key_roles()
        self.add_scores()
        return self.individuals_df

```
